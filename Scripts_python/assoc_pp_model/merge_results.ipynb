```python
name_dataset = "LUCAS_SOC_all_26650_NocitaKS"
type_optim = "progressive_optim"
```

## Merge Performance results

```python
# Charger les deux fichiers CSV
path_name = f"Results/assoc_pp_model/per_dataset/{name_dataset}/"
df1 = pd.read_csv(path_name + f"results_{name_dataset}_{type_optim}_LGBM.csv")  # contient 1 modèle
df2 = pd.read_csv(path_name + f"results_{name_dataset}_{type_optim}_10000_epc_1000_ptc_Ridge_PLS_NICON.csv")  # contient 3 modèles

# Concaténer verticalement
df_merged = pd.concat([df1, df2], ignore_index=True)

# Sauvegarder dans un nouveau fichier CSV
df_merged.to_csv(path_name + f"results_{name_dataset}_{type_optim}_10000_epc_1000_ptc.csv", index=False)
```

## Merge Timing results per model

```python
# Charger les deux fichiers CSV
path_name = f"Results/assoc_pp_model/per_dataset/{name_dataset}/"
df1 = pd.read_csv(path_name + f"timing_per_model_{type_optim}_LGBM.csv")  # contient 1 modèle
df2 = pd.read_csv(path_name + f"timing_per_model_10000epochs_1000patience_{type_optim}_Ridge_PLS_NICON.csv")  # contient 3 modèles

# Concaténer verticalement
df_merged = pd.concat([df1, df2], ignore_index=True)

# Sauvegarder dans un nouveau fichier CSV
df_merged.to_csv(path_name + f"timing_per_model_{type_optim}_10000_epc_1000_ptc.csv", index=False)
```

## Merge global timing results

```python
# Load the two csv files
path_name = f"Results/assoc_pp_model/per_dataset/{name_dataset}/"
df1 = pd.read_csv(path_name + f"timing_results_{type_optim}_LGBM.csv")  # contient 1 modèle
df2 = pd.read_csv(path_name + f"timing_results_10000epochs_1000patience_{type_optim}_Ridge_PLS_NICON.csv")  # contient 3 modèles

# Ponderate by the number of models in each file divided by the total number of models
time = (df1["time"] + df2['time'])

# Create a dataframe with the appropriate time value
df = df1.copy()
df['time'] = time

# Save the new csv file
df.to_csv(path_name + f"timing_results_10000epochs_1000patience_{type_optim}.csv", index=False)
```
